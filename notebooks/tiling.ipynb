```python
import os
import cv2
from tqdm import tqdm

tile_size = 480
overlap_ratio = 0.5
stride = int(tile_size * (1 - overlap_ratio))

image_dir = "/content/c2a-dataset/new_dataset3/train/images"
label_dir = "/content/c2a-dataset/new_dataset3/train/labels"
output_img_dir = "/content/tiling-dataset/train/images"
output_lbl_dir = "/content/tiling-dataset/train/labels"

os.makedirs(output_img_dir, exist_ok=True)
os.makedirs(output_lbl_dir, exist_ok=True)

def read_yolo_labels(txt_path, img_w, img_h):
    labels = []
    if not os.path.exists(txt_path):
        return labels
    with open(txt_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            cls, xc, yc, w, h = map(float, parts)
            abs_w = w * img_w
            abs_h = h * img_h
            abs_xc = xc * img_w
            abs_yc = yc * img_h
            x1 = abs_xc - abs_w / 2
            y1 = abs_yc - abs_h / 2
            x2 = abs_xc + abs_w / 2
            y2 = abs_yc + abs_h / 2
            labels.append((int(cls), x1, y1, x2, y2))
    return labels

tile_id = 1
image_files = [f for f in os.listdir(image_dir) if f.lower().endswith(('.jpg', '.png'))]

for img_name in tqdm(image_files, desc="🔄 Đang xử lý ảnh", leave=True):
    img_path = os.path.join(image_dir, img_name)
    label_path = os.path.join(label_dir, os.path.splitext(img_name)[0] + ".txt")

    img = cv2.imread(img_path)
    if img is None:
        print(f"Bỏ qua ảnh lỗi: {img_path}")
        continue

    img_h, img_w = img.shape[:2]
    labels = read_yolo_labels(label_path, img_w, img_h)

    if img_h <= tile_size and img_w <= tile_size:
        tile_img_path = os.path.join(output_img_dir, f"image_{tile_id}.jpg")
        tile_lbl_path = os.path.join(output_lbl_dir, f"image_{tile_id}.txt")
        cv2.imwrite(tile_img_path, img)

        if labels:
            new_labels = []
            for cls, x1, y1, x2, y2 in labels:
                new_xc = (x1 + x2) / 2 / img_w
                new_yc = (y1 + y2) / 2 / img_h
                new_w = (x2 - x1) / img_w
                new_h = (y2 - y1) / img_h
                new_labels.append(f"{cls} {new_xc:.6f} {new_yc:.6f} {new_w:.6f} {new_h:.6f}")

            with open(tile_lbl_path, 'w') as f:
                f.write('\n'.join(new_labels))
        tile_id += 1
        continue

    y_steps = set(range(0, img_h - tile_size + 1, stride))
    y_steps.add(img_h - tile_size)
    x_steps = set(range(0, img_w - tile_size + 1, stride))
    x_steps.add(img_w - tile_size)

    for y in sorted(list(y_steps)):
        for x in sorted(list(x_steps)):
            if y < 0 or x < 0: continue

            tile = img[y:y + tile_size, x:x + tile_size]

            tile_img_path = os.path.join(output_img_dir, f"image_{tile_id}.jpg")
            tile_lbl_path = os.path.join(output_lbl_dir, f"image_{tile_id}.txt")

            cv2.imwrite(tile_img_path, tile)

            new_labels = []
            for cls, x1_orig, y1_orig, x2_orig, y2_orig in labels:
                inter_x1 = max(x1_orig, x)
                inter_y1 = max(y1_orig, y)
                inter_x2 = min(x2_orig, x + tile_size)
                inter_y2 = min(y2_orig, y + tile_size)

                if inter_x1 >= inter_x2 or inter_y1 >= inter_y2:
                    continue

                new_x1 = inter_x1 - x
                new_y1 = inter_y1 - y
                new_x2 = inter_x2 - x
                new_y2 = inter_y2 - y

                new_xc = (new_x1 + new_x2) / 2 / tile_size
                new_yc = (new_y1 + new_y2) / 2 / tile_size
                new_w = (new_x2 - new_x1) / tile_size
                new_h = (new_y2 - new_y1) / tile_size

                if new_w < 0.01 or new_h < 0.01:
                     continue

                # if new_w < 0.03 or new_h < 0.03 or (new_w * new_h) < 0.001:
                #     continue

                new_labels.append(f"{cls} {new_xc:.6f} {new_yc:.6f} {new_w:.6f} {new_h:.6f}")

            with open(tile_lbl_path, 'w') as f:
                f.write('\n'.join(new_labels))

            tile_id += 1

print(f"✅ Đã xử lý {len(image_files)} ảnh → Tổng số tile đã tạo: {tile_id - 1}")
```

```python
import os

def remove_jpg_without_labels(image_dir, label_dir):
    removed = 0
    for img_file in os.listdir(image_dir):
        if not img_file.lower().endswith(".jpg"):
            continue

        img_path = os.path.join(image_dir, img_file)
        label_name = os.path.splitext(img_file)[0] + ".txt"
        label_path = os.path.join(label_dir, label_name)

        if not os.path.exists(label_path):
            os.remove(img_path)
            removed += 1

    print(f"🧹 Đã xoá {removed} ảnh .jpg không có label.")
```

```python
def draw_bboxes_on_image(image_path, label_path, output_path):
    img = cv2.imread(image_path)
    if img is None:
        print(f"Không đọc được ảnh: {image_path}")
        return

    img_h, img_w = img.shape[:2]

    with open(label_path, 'r') as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) != 5:
                continue
            cls, xc, yc, w, h = map(float, parts)

            xc *= img_w
            yc *= img_h
            w *= img_w
            h *= img_h

            x1 = int(xc - w/2)
            y1 = int(yc - h/2)
            x2 = int(xc + w/2)
            y2 = int(yc + h/2)

            color = (0, 255, 0)
            thickness = 2
            cv2.rectangle(img, (x1, y1), (x2, y2), color, thickness)

            font = cv2.FONT_HERSHEY_SIMPLEX
            font_scale = 0.5
            font_thickness = 1
            text_color = (255, 255, 255)
            cv2.putText(img, str(int(cls)), (x1, y1 - 5), font, font_scale, text_color, font_thickness, cv2.LINE_AA)


    cv2.imwrite(output_path, img)
    print(f"Đã lưu ảnh với bounding box tại: {output_path}")

output_image_with_bboxes = "/content/test.jpg"
input_image = "/content/tiling-dataset/train/images/image_1032.jpg"
input_label = "/content/tiling-dataset/train/labels/image_1032.txt"
draw_bboxes_on_image(input_image, input_label, output_image_with_bboxes)

# --- Tùy chọn: Sử dụng hàm để vẽ bounding box cho từng tile đã tạo ---
# for y in range(0, img_h - tile_size + 1, stride):
#     for x in range(0, img_w - tile_size + 1, stride):
#         tile_name = f"tile_{tile_id}" # Cần tính lại tile_id trong vòng lặp nếu muốn chạy độc lập
#         tile_img_path = os.path.join(output_dir, tile_name + ".jpg")
#         tile_lbl_path = os.path.join(output_dir, tile_name + ".txt")
#         tile_output_path = os.path.join(output_dir, tile_name + "_with_bboxes.jpg")

#         if os.path.exists(tile_lbl_path): # Chỉ vẽ nếu có file label
#              draw_bboxes_on_image(tile_img_path, tile_lbl_path, tile_output_path)
#         else:
#             # Nếu không có label, vẫn có thể copy ảnh gốc tile
#             import shutil
#             shutil.copy(tile_img_path, tile_output_path)

```
